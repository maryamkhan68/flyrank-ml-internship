## **1) My lane as an ML task (type)**

This is a **ranking problem**.

The goal is to rank content pages based on priority for improvement.

## **2) Target or Proxy**

The target is content performance, approximated using engagement signals like CTR.

This acts as a proxy for identifying which pages need improvement.

## **3) Success metric**

The success metric is **Precision@K**, which measures how well the model identifies the top pages needing improvement.

## **4) Unit of analysis**

Each row represents a **single content item (page/article)**.

The dataset contains performance metrics (such as CTR, engagement, and position) for each content piece over a recent time window.
