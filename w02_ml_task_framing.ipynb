## Problem framing (decision-first)

**What decision does this improve?**  
This improves the decision of which content pages should be prioritized for improvement.

**Who acts on the output, and what do they do?**  
A content editor or SEO manager will use the output to decide which pages to update first.

**What does a wrong answer cost?**  
Wrong decisions may waste time on low-priority pages or miss important declining content.

**Why does ML help here instead of a fixed rule?**  
Content performance depends on multiple interacting factors such as CTR, engagement, and position.

## 1) My lane as an ML task (type)

This is a **ranking problem**.

The goal is to rank content pages based on priority for improvement.

## 2) Target or Proxy

The target is content performance, approximated using engagement signals like CTR.

This acts as a proxy for identifying which pages need improvement.

## 3) Success metric

The success metric is **Precision@K**.
