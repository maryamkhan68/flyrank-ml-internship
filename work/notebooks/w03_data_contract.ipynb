```python
#I verify the data grain, row counts, date range, and availability using queries on March 2026 data.
```

```python
# Grain check
query1 = """
SELECT COUNT(*) AS total_rows,
       COUNT(DISTINCT query, url, date) AS unique_rows
FROM your_table_name
WHERE month = '2026-03';
"""
print(query1)

# Row count + date range
query2 = """
SELECT COUNT(*) AS row_count,
       MIN(date) AS start_date,
       MAX(date) AS end_date
FROM your_table_name
WHERE month = '2026-03';
"""
print(query2)

# Availability check
query3 = """
SELECT COUNT(*) AS total_rows,
       COUNT(*) FILTER (WHERE clicks IS TRUE) AS available_rows
FROM your_table_name
WHERE month = '2026-03';
"""
print(query3)
```
